## Zad 1.
Rozważ proces Wienera ($W(t)$) na przedziale ($[t_0,t_1]$).
Zdefiniuj **warunkowy proces Wienera** (mostek Browna) dany warunkami brzegowymi:
$$
W(t_0)=x_0,\qquad W(t_1)=x_1.
$$

1. Wyprowadź dyskretną postać kowariancji procesu warunkowego na siatce czasowej.
2. Zaimplementuj trzy niezależne metody generowania trajektorii:

   * metodę macierzową z rozkładem Choleskiego - generowanie całej trajektorii jednocześnie przez rozkład macierzy kowariancji,
   * rekurencyjną metodę Brownian bridge - rekurencyjne uzupełnianie punktów pośrednich,
   * metodę szeregową opartą na rozwinięciu Karhunena-Loève - aproksymacja za pomocą szeregu funkcyjnego.
3. Porównaj własności numeryczne metod (złożoność, stabilność, możliwość skalowania).

## Zad 2.
Rozważ uogólniony proces Wienera:
$$
X(t) = x_0 + \mu t + \sigma W(t)
$$
gdzie:
- $x_0$ - wartość początkowa
- $\mu$ - parametr dryfu
- $\sigma$ - parametr zmienności (dyfuzji)

Zaimplementuj symulację tego procesu i zbadaj wpływ parametrów $\mu$ i $\sigma$ na:
- Średnią trajektorię
- Przedziały ufności (np. 95%)
- Prawdopodobieństwo osiągnięcia określonego poziomu
